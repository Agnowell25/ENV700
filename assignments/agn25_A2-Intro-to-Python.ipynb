# Part 6: Functions (15 points)


### A. Function with One Input
- Create a function called `circle_area()`.
- The function should accept a value representing the radius of a circle and return the area of a circle with that radus (`area = 2πr`).
- Challenge: Use the math package to get the value of π. (Otherwise use `3.14`.)

### B. Function with Two Inputs
- Create a function called `rectangle_area()`.
- The function should accept length and width and return area.

### C. Function with Two Outputs
- Create a function called `min_max_values()`.
- The function should accept a list of numbers.
- Return both the minimum and maximum values.

Demonstrate each function with example inputs.


````python
# Write your solution here
import math
# A (Is this circumference rather than area?)
def circle_area(radius):
    area = radius * 2 * math.pi
    return area

print(circle_area(3))

# B 

def rectangle_area(length, width):
    area = length * width
    return area

print(rectangle_area(2, 4))

# C

def min_max_values(values):
    minimum = min(values)
    maximum = max(values)
    return minimum, maximum

numbers = [1, 3, 7, 2, 4]

print(min_max_values(numbers))
````
